where $R_t$ is the rate firms and households pay, $R^{ff}_t$ the fed funds rate, $\bar{f}$ the risk premium and $\bar{a} = -2$ the housing bust.

**Takeaway.** The Fed cuts its rate just enough to offset the housing bust, but firms borrow at that rate plus the premium, so each point of friction costs $\bar{b}$ points of output, exactly like a fall in demand. ### Fig. 3: The financial crisis in the AS/AD framework

*Move the slider through the periods after the shock; watch the AS curve shift down and the economy slide along AD′ toward C.*

$$
\tilde{Y}_t = \bar{a} - \bar{b}\bar{f} - \bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

$$
\pi_t = \pi_{t-1} + \bar{\nu}\,\tilde{Y}_t
$$

$$
\pi_C = \bar{\pi} + \frac{\bar{a} - \bar{b}\bar{f}}{\bar{b}\bar{m}}
$$

where $\bar{a} = -2$ is the housing bust, $\bar{f} = 4$ the risk premium, $\bar{m}$ the policy response, $\bar{\nu}$ the slope of the Phillips curve and $\pi_C$ long-run inflation.

```python
#| echo: false
# Model app: Fig. 3: The financial crisis in the AS/AD framework
```

**Takeaway.** The housing bust and the risk premium both shift the AD curve down, so in period 1 the economy jumps from A to B, with output and inflation both lower. Firms then expect last period's lower inflation, the AS curve shifts down period after period, and the economy slides along AD′ toward C: output recovers to potential, but inflation keeps falling and ends below zero, in deflation.

**Source.** Adapted from Fig. 14.3 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## 14.8 The Great Depression in the short-run model

### Fig. 12: The Great Depression in the IS/MP framework

*Move the slider to add one shock at a time, from A to D; watch which curve shifts and read output in the title.*

$$

$$
R_t = i_t - \pi_t
$$

where $R_t$ is the real interest rate, $i_t$ the nominal rate, which the Fed raises and then holds fixed, $\pi_t$ inflation and $\bar{a}$ the demand shock.

**Takeaway.** Each step pushes output further below potential: the Fed's hike raises the real rate (A to B), and the stock market crash shifts the IS curve back (B to C). The slump then lowers inflation, and with the nominal rate held fixed the real rate rises again (C to D), which deepens the slump and lowers inflation once more: a deflationary spiral.

### Fig. 13: The Phillips curve and the Great Depression

*Move the slider from A to D; each point's height is how much inflation changes in a year, and the title reads it.*

$$
\Delta\pi_t = \bar{\nu}\,\tilde{Y}_t
$$

$$
\tilde{Y}_t = (1 + \bar{b}\bar{\nu})\,\tilde{Y}_{t-1}
$$

where $\Delta\pi_t$ is the change in inflation and $\bar{\nu}$ the slope of the Phillips curve; the second line holds while the nominal rate stays fixed.

**Takeaway.** Inflation falls whenever output is below potential, and by more the deeper the slump, so the fall grows from B to C to D. Starting from zero inflation, that means deflation, and with the nominal rate fixed each point of deflation raises the real rate by a point, so the slump grows by the factor $1 + \bar{b}\bar{\nu}$ each year.

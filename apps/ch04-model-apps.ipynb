## The production function and its equilibrium

### Fig. 1: The diminishing marginal product of capital

*Move the slider to change the number of workers; watch the whole curve shift, and compare the two output gains read in the title.*

$$
Y = \bar{A}K^{1/3}L^{2/3}
$$

$$
MPK = \frac{1}{3}\bar{A}\left(\frac{L}{K}\right)^{2/3}
$$

where $\bar{A} = 1$ is productivity, $L$ the number of workers, set by the slider, and $MPK$ the slope of the curve.

**Takeaway.** With the workforce fixed, each new machine has fewer hands to run it, so it adds less output than the one before: the curve keeps rising but flattens. More workers lift the whole curve, because every machine has more people to run it, and fewer workers lower it. Whatever the number of workers, the eighth machine adds about a third as much as the second.

### Fig. 2: Supply and demand for capital and labor

*Move the slider to shift the supply of capital; watch the rental price in panel (a), the wage in panel (b), and the shares in the title.*

$$
r^* = \frac{1}{3}\bar{A}\left(\frac{\bar{L}}{\bar{K}}\right)^{2/3}
$$

$$
w^* = \frac{2}{3}\bar{A}\left(\frac{\bar{K}}{\bar{L}}\right)^{1/3}
$$

$$
\frac{r^*K^*}{Y^*} = \frac{1}{3}, \qquad \frac{w^*L^*}{Y^*} = \frac{2}{3}
$$

where $\bar{K}$ is the supply of capital, set by the slider, and $\bar{L} = \bar{A} = 1$.

**Takeaway.** A larger supply of capital moves the economy down the capital demand curve, so the rental price falls, and it gives each worker more machines, so labor demand shifts up and the wage rises; a smaller supply does the opposite. Whatever the supply of capital, capital earns one third of output and labor two thirds, because the factor shares equal the exponents of the production function.

**Source.** Adapted from Fig. 4.2 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## Comparing the model with the data

### Fig. 5: The model's prediction for per capita GDP

*Move the slider to change the capital exponent $\alpha$; watch the points against the 45-degree line, and read the share above it in the title.*

$$
Y = \bar{A}K^{\alpha}L^{1-\alpha}
$$

$$
y^* = \bar{k}^{\alpha}
$$

where $y^*$ is predicted income and $\bar{k}$ capital per person, both relative to the United States, with the same $\bar{A}$ in every country.

**Takeaway.** With equal productivity everywhere, the model predicts $y^* = \bar{k}^{\alpha}$, so a larger exponent turns the same capital gaps into larger predicted income gaps. At the textbook's one third, and at every exponent below 1, most countries lie above the 45-degree line, the poorest farthest: the model predicts them much richer than they are. An exponent of 1 overshoots, putting most poor countries below the line, and it clashes with a capital share of about one third in the data.

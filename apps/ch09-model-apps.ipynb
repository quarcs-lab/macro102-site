## The Phillips curve

### Fig. 6: The Phillips curve

*Move the slider to change short-run output; the black dot slides along the line, and the title reads the change in inflation.*

$$
\Delta\pi_t \equiv \pi_t - \pi_{t-1} = \bar{\nu}\,\tilde{Y}_t
$$

where $\pi_t$ is inflation, $\tilde{Y}_t$ short-run output in percent, and $\bar{\nu} = 0.33$ the slope of the curve.

```python
#| echo: false
# Model app: Fig. 6: The Phillips curve
```

**Takeaway.** The Phillips curve turns short-run output into a change in inflation: when output is above potential, firms raise prices faster than everyone expected and inflation rises; when it is below, inflation falls. Only at potential, where the dot sits at the origin, does inflation hold steady, at whatever level it already has. The change is proportional to the gap, so a slump twice as deep lowers inflation twice as fast.

**Source.** Adapted from Fig. 9.6 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## Overstimulating the economy

$$

$$
\pi_t = \pi_0 + \bar{\nu}\,\tilde{Y}\,t
$$

where output is held at $\tilde{Y} = +3$ percent every year and inflation starts at $\pi_0 = \bar{\pi} = 2$ percent.

**Takeaway.** Holding output above potential raises inflation not once but every year, by the same amount, so inflation climbs in a straight line without end. A steeper Phillips curve makes the climb faster and a flatter one slower, but no slope stops it: inflation stops rising only when output returns to potential.

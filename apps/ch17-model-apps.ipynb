## How firms choose their capital stock

### Fig. 1: How much should a firm invest?

*Move the slider to change the interest rate; watch the flat user-cost line shift and the desired capital stock slide along the MPK curve.*

$$
MPK = \frac{1}{3}\,\bar{A}\left(\frac{L}{K}\right)^{2/3}
$$

$$
MPK = R + \bar{d} - \frac{\Delta p_k}{p_k} = uc
$$

$$
K^{\text{desired}} = L\left(\frac{\bar{A}}{3\,uc}\right)^{3/2}
$$

where $uc$ is the user cost of capital, $R$ the interest rate, $\bar{d} = 0.08$ the depreciation rate, $\Delta p_k/p_k$ the capital gain (zero here) and $\bar{A}$ productivity.

**Takeaway.** The firm adds capital as long as one more unit earns more than it costs, so it stops where the falling MPK curve meets the flat user-cost line. A higher interest rate raises the user cost point for point and moves the desired capital stock to the left, and a lower one moves it to the right; because the curve flattens out, each point of interest moves capital more when the user cost is low.

### Fig. 2: An increase in the corporate income tax

*Move the slider to set the corporate tax rate; the dashed line is the old user cost, and the title reads the fall in desired capital.*

$$
MPK = \frac{R + \bar{d} - \Delta p_k/p_k}{1-\tau}
$$

$$
K^{\text{desired}} = L\left(\frac{\bar{A}}{3\,uc}\right)^{3/2}
$$

where $\tau$ is the corporate tax rate and $uc$ the user cost with the tax, the right side of the first equation; without the tax it is 0.10.

**Takeaway.** The government takes a share of what each unit of capital earns but shares none of its costs, so the tax divides the user cost by $1-\tau$ and shifts the user-cost line up. The firm moves up its MPK curve to a smaller capital stock, which falls by a larger percentage than the tax rate itself; at a zero tax rate the two lines coincide and nothing changes.

**Source.** Adapted from Fig. 17.2 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## The stock market

### Fig. 3: The model's price-earnings ratio against the U.S. stock market

*Move the slider to change expected dividend growth in the model; compare the teal model line with the data and the dashed long-run average.*

$$
p_s = \frac{\text{dividend}}{R - g}
$$

$$
\frac{p_s}{\text{earnings}} = \frac{\text{dividend}/\text{earnings}}{R - g}
$$

where $p_s$ is the share price, $R = 0.05$ the interest rate, $g$ the growth rate of dividends, and the payout ratio $\text{dividend}/\text{earnings}$ is 0.5.

**Takeaway.** A share is worth the present value of its dividends, so the model's P/E ratio is the payout ratio divided by the gap $R - g$: faster dividend growth raises it, slowly at first and then steeply. Growth of about 2 percent puts the line near the long-run average, but only growth close to the 5 percent interest rate justifies peaks like those of 1929, 1999 and the mid-2020s, which is why a bubble is hard to tell from a change in expected growth.

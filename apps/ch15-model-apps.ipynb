## The labor market in the DSGE model

### Fig. 2: The labor market in the DSGE model

*Move the slider to change consumption, which shifts only labor supply; watch where the curves cross, and read employment and the wage in the title.*

$$
w = \frac{2}{3}\cdot\frac{\bar{A}K^{1/3}}{L^{1/3}}
$$

$$
L^s = \bar{\ell}\cdot\frac{w}{c}
$$

$$
L^* = \left[\frac{2}{3}\cdot\frac{\bar{A}K^{1/3}\,\bar{\ell}}{c}\right]^{3/4}
$$

where $w$ is the real wage, $L$ employment, $c$ consumption per person, $\bar{\ell}$ the labor-supply scale, and TFP $\bar{A}$ and capital $K$ are both 1.

```python
#| echo: false
# Model app: Fig. 2: The labor market in the DSGE model
```

**Takeaway.** Firms hire until the marginal product of labor equals the wage, and households work more when the wage is high relative to their consumption. A higher $c$ makes extra income worth less to people, so labor supply shifts up and to the left: employment and output fall while the real wage rises a little. A lower $c$ does the opposite, and labor demand never moves.

**Source.** Adapted from Fig. 15.2 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## Real shocks: productivity, taxes and government purchases

### Fig. 3: A negative TFP shock in the DSGE model

*Move the slider to set TFP after the shock; the dashed curve is the old labor demand, and the title reads employment and the real wage.*

$$
w = \frac{2}{3}\cdot\frac{\bar{A}'K^{1/3}}{L^{1/3}}
$$

$$
L^* = \left[\frac{2}{3}\cdot\frac{\bar{A}'K^{1/3}\,\bar{\ell}}{c}\right]^{3/4}, \qquad w^* = \frac{c\,L^*}{\bar{\ell}}
$$

where $\bar{A}'$ is TFP after the shock (1 before it), $c = 0.6$ consumption and $\bar{\ell} = 0.9$ the labor-supply scale.

```python
#| echo: false
# Model app: Fig. 3: A negative TFP shock in the DSGE model
```

**Takeaway.** TFP enters only labor demand, and a temporary shock barely moves consumption, so only the demand curve shifts. A fall in TFP shifts it down, and employment and the real wage both fall by the same percentage, as in a recession; a rise above 1 raises both, like a boom, and at 1 nothing changes.

**Source.** Adapted from Fig. 15.3 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 4: The imposition of a sales tax in the DSGE model

*Move the slider to raise the sales tax; the dashed curve is labor demand before the tax, and the title reads employment and the real wage.*

$$
w = (1-\tau)\cdot\frac{2}{3}\cdot\frac{\bar{A}K^{1/3}}{L^{1/3}}
$$

$$
L^* = \left[(1-\tau)\cdot\frac{2}{3}\cdot\frac{\bar{A}K^{1/3}\,\bar{\ell}}{c}\right]^{3/4}
$$

where $\tau$ is the tax on each dollar of sales and TFP $\bar{A}$ stays at 1.

```python
#| echo: false
# Model app: Fig. 4: The imposition of a sales tax in the DSGE model
```

**Takeaway.** Firms keep only part of what an extra worker produces, so the tax shifts labor demand down exactly as a fall in TFP would: only the product $(1-\tau)\bar{A}$ enters the equilibrium. Employment and the real wage both fall, by more the higher the tax, and with no tax nothing changes.

**Source.** Adapted from Fig. 15.4 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 5: A rise in government purchases

*Move the slider to set consumption after the rise in purchases; the dashed line is the old labor supply, and the arrows show the move.*

$$
L^s = \bar{\ell}\cdot\frac{w}{c'}
$$

$$
L^* = \left[\frac{2}{3}\cdot\frac{\bar{A}K^{1/3}\,\bar{\ell}}{c'}\right]^{3/4}, \qquad w^* = \frac{c'\,L^*}{\bar{\ell}}
$$

where $c'$ is consumption after the rise in purchases, 0.6 before it.

```python
#| echo: false
# Model app: Fig. 5: A rise in government purchases
```

**Takeaway.** Government purchases enter neither curve; they reach the labor market through consumption, because households expect higher taxes and feel poorer. A lower $c'$ shifts labor supply down and to the right, so employment rises while the real wage falls, by much less: a boom with falling wages and consumption, unlike a normal business cycle. At 0.6 nothing changes.

**Source.** Adapted from Fig. 15.5 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## Monetary policy with sticky wages and sticky prices

### Fig. 6: Sticky wages in the DSGE model

*Move the slider to change the sticky nominal wage; the bracket marks unemployment, the gap between labor supply and labor demand at that wage.*

$$
L = \min\left\{L^d\!\left(\frac{\bar{W}}{P}\right),\ L^s\!\left(\frac{\bar{W}}{P}\right)\right\}
$$

$$
L^d = \left(\frac{2}{3}\cdot\frac{\bar{A}K^{1/3}}{w}\right)^{3}, \qquad L^s = \bar{\ell}\cdot\frac{w}{c}
$$

where $\bar{W}$ is the sticky nominal wage, $P = 1$ the price level and $w = \bar{W}/P$ the real wage.

```python
#| echo: false
# Model app: Fig. 6: Sticky wages in the DSGE model
```

**Takeaway.** The sticky real wage is above the market-clearing level of about 0.67, so households want to work more than firms want to hire, and the short side of the market, the firms, decides employment. The gap is involuntary unemployment, and a higher $\bar{W}$ widens it from both sides: firms hire fewer workers while more people look for a job.

**Source.** Adapted from Fig. 15.6 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 7: A monetary expansion with sticky wages

*Move the slider through the periods: prices rise in period 1 and the nominal wage adjusts in period 2; watch the unemployment bracket.*

$$
L = \min\left\{L^d\!\left(\frac{\bar{W}}{P}\right),\ L^s\!\left(\frac{\bar{W}}{P}\right)\right\}
$$

$$
\ln L^d = \text{const} - 3\ln\bar{W} + 3\ln P
$$

where $\bar{W} = 0.75$ is the sticky nominal wage and the price level rises from $P_0 = 1$ to $P' = 1.05$.

```python
#| echo: false
# Model app: Fig. 7: A monetary expansion with sticky wages
```

**Takeaway.** With the nominal wage stuck, higher prices lower the real wage, so firms hire more and unemployment shrinks, but the real wage and employment move in opposite directions, unlike in actual booms. Once the nominal wage adjusts, the real wage returns to its market-clearing level and unemployment disappears: in the long run, money does not change employment.

**Source.** Adapted from Fig. 15.7 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 8: A monetary expansion with sticky prices

*Move the slider to change the output demanded; the vertical line is labor demand and the dashed one its old position.*

$$
L = \left(\frac{Y'}{\bar{A}K^{1/3}}\right)^{3/2}
$$

$$
\frac{W}{P} = \frac{c\,L}{\bar{\ell}}
$$

where $Y'$ is the output demanded (1 before the shock), $W/P$ the real wage and $c = 0.6$ consumption.

```python
#| echo: false
# Model app: Fig. 8: A monetary expansion with sticky prices
```

**Takeaway.** With prices stuck, firms produce whatever is demanded and hire exactly the labor that output needs, so labor demand is a vertical line. More output demanded shifts it right, and the real wage must rise along labor supply to attract the extra workers: employment and the real wage rise together, as in actual booms. Less output demanded moves both down, and at 1 nothing changes.

**Source.** Adapted from Fig. 15.8 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## Good news about the future

$$
L^* = \left[\frac{2}{3}\cdot\frac{\bar{A}K^{1/3}\,\bar{\ell}}{c'}\right]^{3/4}, \qquad w^* = \frac{c'\,L^*}{\bar{\ell}}
$$

where $c'$ is consumption after the news, 0.6 before it.

```python
#| echo: false
# Model app: Fig. 9: Good news in the DSGE model: flexible prices
```

**Takeaway.** The news changes nothing in today's production, so labor demand stays put; it works only through consumption, the model's forward-looking variable. Feeling richer, households consume more and want more leisure, so a higher $c'$ shifts labor supply up and to the left: employment falls and the real wage rises a little. Bad news, below 0.6, does the opposite, and at 0.6 nothing changes.

where the first line holds with flexible prices and the second with sticky prices; consumption rises from 0.6 to $c' = 0.7$, and with sticky prices the output demanded from 1 to $Y' = 1.1$.

**Takeaway.** The same rise in consumption shifts labor supply up in both cases, but labor demand differs.

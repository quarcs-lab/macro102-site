**Takeaway.** With constant velocity, money that grows faster than real GDP pushes up prices one for one, so the model predicts that economies lie along the line $\pi = \bar{g}_M - \bar{g}_Y$. Faster real GDP growth shifts the line down, most visibly at low money growth on the ratio scale; at zero growth it is the 45-degree line. The economies do line up along a rising band, as predicted, but it takes real GDP growth of about 5 to 6 percent, more than the few percent a year it usually grows, to run the line through the middle of the cloud: a sign that velocity fell in many economies.

## The mathematics of growth

### Fig. 3: Population over time

*Move the slider to change the growth rate; watch how sharply the curve bends upward, and read the population after 100 years in the title.*

$$
L_{t+1} = L_t\,(1+\bar{n})
$$

$$
L_t = L_0\,(1+\bar{n})^t
$$

where $L_t$ is the population in year $t$, $L_0 = 6$ billion, and $\bar{n}$ is the constant growth rate.

**Takeaway.** Each year the population is multiplied by the same growth factor $1+\bar{n}$, so the yearly increase is a fixed share of an ever larger base and the curve bends upward. A higher growth rate makes it bend more sharply, and small differences in the rate become large differences in the population after a century; at a zero rate the line stays flat.

**Source.** Adapted from Fig. 3.3 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 4: Population over time, revisited

*Move the slider to change the growth rate; watch the slope of the straight line in panel (b), and read the doubling time in the title.*

$$
\ln L_t = \ln L_0 + t\,\ln(1+\bar{n})
$$

$$
t_{\text{double}} = \frac{\ln 2}{\ln(1+\bar{n})} \approx \frac{70}{100\,\bar{n}}
$$

where $L_t$ is the population in year $t$, $L_0 = 6$ billion, and $100\,\bar{n}$ is the growth rate in percent.

**Takeaway.** On a ratio scale equal vertical steps are equal percentage changes, so constant growth draws a straight line whose slope is the growth rate, with the doublings evenly spaced. A higher growth rate makes the line steeper and the doubling time shorter, and a lower one does the opposite. At every rate shown, the Rule of 70 is within half a year of the exact doubling time.

**Source.** Adapted from Fig. 3.4 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 5: Constant growth against U.S. per capita GDP on a ratio scale

$$

$$
\ln y_t = \ln y_0 + t\,\ln(1+\bar{g})
$$

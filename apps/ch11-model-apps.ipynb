## 11.3 Deriving the IS curve

### Fig. 1: Introducing the IS curve

*Move the slider to change the real interest rate; the dot slides along the IS curve, and the title reads investment and output.*

$$
\frac{I_t}{\bar{Y}_t} = \bar{a}_i - \bar{b}\,(R_t - \bar{r})
$$

$$
\tilde{Y}_t = \bar{a} - \bar{b}\,(R_t - \bar{r})
$$

where $R_t$ is the real interest rate, $\bar{r} = 5$ percent the marginal product of capital, $\bar{a}_i$ the normal investment share, $\bar{a} = 0$ and $\bar{b} = 2$.

```python
#| echo: false
# Model app: Fig. 1: Introducing the IS curve
```

**Takeaway.** A higher real interest rate makes borrowing dearer, so firms invest less and output falls: each point on the rate lowers investment, and output, by $\bar{b}$ points of potential output. At $R_t = \bar{r}$ investment is at its normal share and output at potential; above it output is below potential, and below it output is above.

**Source.** Adapted from Fig. 11.1 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 2: The IS curve

*Move the slider to change the sensitivity of investment; the dashed line is the baseline curve, and watch the new one turn around the dot.*

$$
\tilde{Y}_t = \bar{a} - \bar{b}\,(R_t - \bar{r})
$$

$$
R_t = \bar{r} + \frac{\bar{a} - \tilde{Y}_t}{\bar{b}}
$$

where $\bar{a} = 0$ is the aggregate demand shock, $\bar{r} = 5$ percent the marginal product of capital and $\bar{b}$ the sensitivity of investment.

```python
#| echo: false
# Model app: Fig. 2: The IS curve
```

**Takeaway.** The sensitivity $\bar{b}$ measures how strongly investment, and so output, reacts to the gap between the real interest rate and the marginal product of capital. A higher value turns the curve around the dot into a flatter line, so the same change in the rate moves output more; a lower value makes it steeper, and at 2 the curve is the baseline. At the dot, where $R_t = \bar{r}$, the interest-rate term vanishes, so output stays at $\bar{a}$ whatever the slope.

**Source.** Adapted from Fig. 11.2 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## 11.4 Using the IS curve

### Fig. 3: An increase in the real interest rate

*Move the slider to set the new interest rate; the economy moves along the curve from A to B, and the title reads output.*

$$
\tilde{Y}_t = \bar{a} - \bar{b}\,(R_t - \bar{r})
$$

where $\bar{a} = 0$, $\bar{b} = 2$ and $\bar{r} = 5$ percent, and the real interest rate rises from $R = \bar{r}$ at A to $R'$ at B.

```python
#| echo: false
# Model app: Fig. 3: An increase in the real interest rate
```

**Takeaway.** Only the interest rate changes, so the IS curve stays where it is and the economy slides up along it, from A to B. Above 5 percent, borrowing costs more than new capital earns, so investment falls and each point on the rate lowers output by $\bar{b} = 2$ points; at 5 percent the economy stays at A.

**Source.** Adapted from Fig. 11.3 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 4: An aggregate demand shock

*Move the slider to set the new demand shock; the dashed line is the old curve, and watch output move from A to B.*

$$
\tilde{Y}_t = \bar{a} - \bar{b}\,(R_t - \bar{r})
$$

$$
\bar{a} \equiv \bar{a}_c + \bar{a}_i + \bar{a}_g + \bar{a}_{ex} - \bar{a}_{im} - 1
$$

where $\bar{a}$ collects the five demand shares and moves to the new shock $\bar{a}'$, while the interest rate stays at $R = \bar{r} = 5$ percent.

```python
#| echo: false
# Model app: Fig. 4: An aggregate demand shock
```

**Takeaway.** A demand shock changes spending at every interest rate, so it shifts the whole IS curve: a positive shock shifts it right, a negative one shifts it left, and at zero nothing moves. With the interest rate still at $\bar{r}$, the interest-rate term is zero, so output moves one for one with the shock, from A to B.

**Source.** Adapted from Fig. 11.4 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## 11.5 Microfoundations: consumption and fiscal policy

### Fig. 5: The life-cycle model of consumption

*Move the slider to change the length of life; the dashed line is consumption, and the title reads its level and the saving years.*

$$
\sum_{t=0}^{T-1}\frac{y^p}{(1+R)^t} = \sum_{t=0}^{T-1}\frac{y_t}{(1+R)^t}
$$

$$
c_t = y^p = \frac{1}{T}\sum_{t=0}^{T-1} y_t
$$

where $y_t$ is income at age $t$, $T$ the length of life, $y^p$ permanent income and $c_t$ consumption, with no interest ($R = 0$).

```python
#| echo: false
# Model app: Fig. 5: The life-cycle model of consumption
```

**Takeaway.** Consumption is set at average lifetime income, so it stays flat while income rises and falls: people borrow or get help when young, save in middle age and live off their savings in retirement. A longer life adds retirement years with low income, so average income, and with it consumption, falls and people must save more in middle age; a shorter life does the reverse.

**Source.** Adapted from Fig. 11.5 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

**Takeaway.** A cut in the real interest rate is again a movement along the IS curve, this time down it. where $\bar{r} = 5$ percent is the old marginal product of capital and $\bar{r}'$ the new one, with $\bar{a} = 0$ and $R = \bar{r}$ held fixed.

**Takeaway.** Better technology makes new capital more productive, so the IS curve shifts up and now reaches potential output at the new, higher $\bar{r}'$.

## What a model does

### Fig. 6: The structure of economic models

*Move the slider to set a minimum wage; watch the outcomes box on the right, where the model computes the wage, employment and unemployment.*

$$
L^s = \bar{a}\,w + \bar{\ell}, \qquad L^d = \bar{f} - w
$$

$$
w = \max(w^*, \bar{w}_{\min}), \qquad L = \bar{f} - w
$$

$$
U = (1+\bar{a})\,(\bar{w}_{\min} - w^*) \quad \text{if } \bar{w}_{\min} > w^*
$$

where $\bar{w}_{\min}$ is the minimum wage, $w^* = 10$ the market wage, and $\bar{a} = 2$, $\bar{\ell} = 30$ and $\bar{f} = 60$ are held fixed.

**Takeaway.** You only choose the inputs; the model's equations work out the wage, employment and unemployment. A minimum wage at or below the market wage $w^*$ changes nothing. Above it, the wage equals the floor, and each extra dollar cuts employment by one hour and raises unemployment by $1 + \bar{a} = 3$ hours.

**Source.** Adapted from Fig. 1.6 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## Solving a model: the labor market

### Fig. 8: The labor market

*Move the slider to change labor demand at a zero wage; the dashed gray lines are the textbook's curves, and the title reads the new equilibrium.*

$$
L^s = \bar{a}\,w + \bar{\ell}, \qquad L^d = \bar{f} - w
$$

$$
w^* = \frac{\bar{f} - \bar{\ell}}{1 + \bar{a}}, \qquad L^* = \frac{\bar{a}\bar{f} + \bar{\ell}}{1 + \bar{a}}
$$

where $\bar{f}$ is the hours firms would hire at a zero wage, with $\bar{a} = 2$ and $\bar{\ell} = 30$ held fixed.

**Takeaway.** A higher $\bar{f}$ shifts labor demand to the right along an unchanged supply curve, so the wage and employment both rise; a lower one shifts it to the left and both fall, and at 60 nothing changes. The shift is shared: the wage takes the share $1/(1+\bar{a})$, one third, and employment the other two thirds.

## 13.3 Monetary policy rules and the AD curve

### Fig. 1: The aggregate demand curve

*Move the slider to change the demand shock; the dashed gray line is the baseline AD curve, and the dot marks where inflation is on target.*

$$
\tilde{Y}_t = \bar{a} - \bar{b}\,(R_t - \bar{r})
$$

$$
R_t - \bar{r} = \bar{m}\,(\pi_t - \bar{\pi})
$$

$$
\tilde{Y}_t = \bar{a} - \bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

where $\tilde{Y}_t$ is short-run output, $\pi_t$ inflation, $R_t$ the real interest rate, $\bar{r}$ the marginal product of capital, $\bar{\pi}$ the inflation target, $\bar{m}$ the policy response, $\bar{b}$ the interest sensitivity and $\bar{a}$ the demand shock.

```python
#| echo: false
# Model app: Fig. 1: The aggregate demand curve
```

**Takeaway.** The AD curve slopes down because of the central bank: when inflation is above the target, the rule raises the real interest rate, spending falls and output drops below potential. A positive demand shock shifts the whole curve to the right by $\bar{a}$, a negative one shifts it to the left, and at zero the curve stays on the baseline.

**Source.** Adapted from Fig. 13.1 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 2: The AD curve after an inflation shock

*Move the slider to set the new inflation rate; the point slides along the AD curve from A to B, and the title follows the chain.*

$$
R_t - \bar{r} = \bar{m}\,(\pi_t - \bar{\pi})
$$

$$
\tilde{Y}_t = \bar{a} - \bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

where $\pi'$ is the new inflation rate; here $\bar{a} = 0$, $\bar{\pi} = 2$ percent and $\bar{b}\bar{m} = 1$.

```python
#| echo: false
# Model app: Fig. 2: The AD curve after an inflation shock
```

**Takeaway.** A change in inflation moves the economy along the AD curve; it does not shift it. Inflation above the 2 percent target makes the rule raise the real interest rate, so output falls below potential; inflation below the target lowers the rate and lifts output above potential, and at 2 percent nothing changes.

**Source.** Adapted from Fig. 13.2 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 3: An aggressive monetary policy rule

*Move the slider to change the policy response; the dashed gray line is the baseline AD curve, and the title reads the recession at B.*

$$
\tilde{Y}_t = \bar{a} - \bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

$$
\pi_t = \bar{\pi} + \frac{\bar{a} - \tilde{Y}_t}{\bar{b}\bar{m}}
$$

where $\bar{m}$ is the policy response, $\bar{b} = 2$ the interest sensitivity, and inflation rises from the 2 percent target to 3 percent.

```python
#| echo: false
# Model app: Fig. 3: An aggressive monetary policy rule
```

**Takeaway.** The policy response sets how much the real interest rate rises for each point of inflation, so it turns the AD curve around A. A response above 0.5 flattens the curve and turns the same rise in inflation into a deeper recession at B; a response below 0.5 steepens it and softens the recession, and at 0.5 it is the baseline curve.

**Source.** Adapted from Fig. 13.3 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## 13.4 Aggregate supply and the AS/AD diagram

### Fig. 4: The aggregate supply curve

*Move the slider to change last year's inflation; the dashed gray line is the baseline AS curve, and the dot marks its height at potential output.*

$$
\pi_t - \pi_{t-1} = \bar{\nu}\tilde{Y}_t + \bar{o}
$$

$$
\pi_t = \pi_{t-1} + \bar{\nu}\tilde{Y}_t + \bar{o}
$$

where $\pi_{t-1}$ is last year's inflation, which firms expect, $\bar{\nu} = 1$ the slope and $\bar{o} = 0$ the inflation shock.

```python
#| echo: false
# Model app: Fig. 4: The aggregate supply curve
```

**Takeaway.** Firms expect last year's inflation, so the AS curve crosses potential output at that rate, and it slopes up because a boom pushes prices up faster than expected. Higher inflation last year shifts the whole curve up one for one, lower inflation shifts it down, and at 2 percent it is the baseline curve.

**Source.** Adapted from Fig. 13.4 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 5: The AS/AD framework

*Move the slider to change expected inflation; watch where the AS and AD curves cross, and read in the title whether that is a steady state.*

$$
\tilde{Y}_t = \bar{a} - \bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

$$
\pi_t = \pi_{t-1} + \bar{\nu}\tilde{Y}_t + \bar{o}
$$

$$
\tilde{Y}^* = 0, \qquad \pi^* = \bar{\pi}
$$

where expected inflation is last year's, $\pi_{t-1}$, and there are no shocks ($\bar{a} = \bar{o} = 0$).

```python
#| echo: false
# Model app: Fig. 5: The AS/AD framework
```

**Takeaway.** When firms expect the 2 percent target, the curves cross at the steady state, with output at potential and inflation on target. Expected inflation above the target shifts the AS curve up, so inflation ends between the target and expected inflation while output falls below potential; expected inflation below the target does the opposite. Off the target the crossing is not a steady state, because inflation differs from what firms expected and the AS curve will move again.

**Source.** Adapted from Fig. 13.5 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## 13.5 Event 1: an inflation shock

### Fig. 6: The initial response to an inflation shock

*Move the slider to change the inflation shock; watch AS jump up to AS₁ and the economy slide down the AD curve to B.*

$$
\pi_t = \pi_{t-1} + \bar{\nu}\tilde{Y}_t + \bar{o}
$$

$$
\pi_1 - \bar{\pi} = \lambda\bar{o}, \qquad \tilde{Y}_1 = -\bar{b}\bar{m}\lambda\,\bar{o}
$$

where $\bar{o}$ is the inflation shock, in period 1 only, and $\lambda = 1/(1 + \bar{b}\bar{m}\bar{\nu})$ = 1/2 with the baseline numbers.

```python
#| echo: false
# Model app: Fig. 6: The initial response to an inflation shock
```

**Takeaway.** The shock shifts the AS curve up by $\bar{o}$, but the policy rule answers the higher inflation with a higher real interest rate, so the economy slides down the AD curve: inflation rises and output falls at the same time, which is stagflation. With the baseline numbers half of the shock shows up as inflation and half as lost output, so a bigger shock brings both more inflation and a deeper recession.

**Source.** Adapted from Fig. 13.6 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 7: Two periods after an inflation shock

*Move the slider to change the policy response; panel (a) shows where AS₂ starts, and panel (b) the move from B to C.*

$$
\pi_2 = \pi_1 + \bar{\nu}\tilde{Y}_2
$$

$$
\pi_2 - \bar{\pi} = \lambda\,(\pi_1 - \bar{\pi}), \qquad \tilde{Y}_2 = -\bar{b}\bar{m}\,(\pi_2 - \bar{\pi})
$$

where $\pi_1$ is inflation in period 1, when the shock $\bar{o} = 2$ hit, and $\lambda = 1/(1 + \bar{b}\bar{m}\bar{\nu})$, which falls as $\bar{m}$ rises.

```python
#| echo: false
# Model app: Fig. 7: Two periods after an inflation shock
```

**Takeaway.** The shock is over, but firms now expect period 1's higher inflation, so AS₂ starts from $\pi_1$, above the original AS curve. From B to C inflation falls and output recovers part of the way, and both gaps shrink by the same factor $\lambda$. A higher policy response means less inflation at B and C and a faster return, at the cost of a deeper recession at B.

**Source.** Adapted from Fig. 13.7 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 8: Period by period after an inflation shock

*Move the slider to step through the periods; watch the AS curve creep back down while the economy slides along the same AD curve toward A.*

$$
\pi_t - \bar{\pi} = \lambda\,(\pi_{t-1} - \bar{\pi})
$$

$$
\tilde{Y}_t = -\bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

where the shock $\bar{o} = 2$ hit in period 1 only, and $\lambda = 1/(1 + \bar{b}\bar{m}\bar{\nu})$ = 1/2 with the baseline numbers.

```python
#| echo: false
# Model app: Fig. 8: Period by period after an inflation shock
```

**Takeaway.** Each period firms expect last period's inflation, so the AS curve shifts down step by step, while the policy rule stays the same and every point lies on the original AD curve. In every period shown inflation is still above the target and output below potential, but both gaps halve each period, so the steps get shorter: the economy moves fastest when it is farthest from the steady state.

**Source.** Adapted from Fig. 13.8 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 9: The effects of an inflation shock: summary

*Move the slider to change the AS slope; panel (b) shows the time paths, and the title reads the inflation peak and the sum of output.*

$$
\pi_t - \bar{\pi} = \lambda\,(\pi_{t-1} - \bar{\pi})
$$

$$
\lambda = \frac{1}{1 + \bar{b}\bar{m}\bar{\nu}}
$$

$$
\sum_{t}\tilde{Y}_t = -\frac{\bar{o}}{\bar{\nu}}
$$

where $\bar{\nu}$ is the slope of the AS curve, $\bar{o} = 2$ the shock in period 1, and the sum runs over the whole episode.

```python
#| echo: false
# Model app: Fig. 9: The effects of an inflation shock: summary
```

**Takeaway.** The shock lasts one period, but it raises expected inflation, so output stays below potential until inflation is back at the target, and over the whole episode it adds up to about $-\bar{o}/\bar{\nu}$. A flatter AS curve means prices react less to the slump, so the recession is deeper, longer and costlier; a steeper curve makes it milder, shorter and cheaper.

**Source.** Adapted from Fig. 13.9 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612) Panel (b), the time paths of inflation and output, is our addition.

## 13.6 Event 2: disinflation

### Fig. 10: The initial response to disinflation

*Move the slider to set the new inflation target; AD shifts down to AD₁, while the AS curve stays where the old expectations put it.*

$$
\tilde{Y}_t = -\bar{b}\bar{m}\,(\pi_t - \bar{\pi}')
$$

$$
\pi_1 - \bar{\pi}' = \lambda\,(\bar{\pi} - \bar{\pi}'), \qquad \tilde{Y}_1 = -\bar{b}\bar{m}\lambda\,(\bar{\pi} - \bar{\pi}')
$$

where $\bar{\pi} = 4$ percent is the old target, $\bar{\pi}'$ the new one, and $\lambda = 1/(1 + \bar{b}\bar{m}\bar{\nu})$ = 1/2.

```python
#| echo: false
# Model app: Fig. 10: The initial response to disinflation
```

**Takeaway.** The target enters only the AD curve, so a lower target shifts AD down by the size of the cut, while the AS curve stays put because firms still expect the old 4 percent. Inflation falls only halfway to the new target and output falls below potential, more so the bigger the cut; with a new target of 4 percent nothing changes.

**Source.** Adapted from Fig. 13.10 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 11: The dynamics of disinflation

*Move the slider to change the policy response; panel (b) shows inflation and output over time, and the title reads the total output lost.*

$$
\pi_t - \bar{\pi}' = \lambda\,(\pi_{t-1} - \bar{\pi}')
$$

$$
\lambda = \frac{1}{1 + \bar{b}\bar{m}\bar{\nu}}
$$

$$
\sum_{t}\tilde{Y}_t = -\frac{\bar{\pi} - \bar{\pi}'}{\bar{\nu}}
$$

where the target falls from $\bar{\pi} = 4$ to $\bar{\pi}' = 2$ percent, $\bar{m}$ is the policy response and $\bar{\nu} = 1$ the slope of the AS curve.

```python
#| echo: false
# Model app: Fig. 11: The dynamics of disinflation
```

**Takeaway.** Firms lower their expectations only period by period, so the AS curve shifts down step by step and the economy slides down AD₁ from B to the new steady state C, with inflation on the new target and output back at potential. A more aggressive central bank makes the recession deeper but shorter, while the total output lost stays at about 2 percent of a year's output, set only by the size of the cut and the slope $\bar{\nu}$.

**Source.** Adapted from Fig. 13.11 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## 13.7 Event 3: a positive aggregate demand shock

### Fig. 12: A positive aggregate demand shock

*Move the slider to change the demand shock; AD shifts to AD₁, and the economy moves along the unchanged AS curve from A to B.*

$$
\tilde{Y}_t = \bar{a} - \bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

$$
\tilde{Y}_1 = \lambda\bar{a}, \qquad \pi_1 - \bar{\pi} = \lambda\bar{\nu}\bar{a}
$$

where $\bar{a}$ is the demand shock, which lasts from period 1 to 9, and $\lambda = 1/(1 + \bar{b}\bar{m}\bar{\nu})$ = 1/2.

```python
#| echo: false
# Model app: Fig. 12: A positive aggregate demand shock
```

**Takeaway.** A demand shock shifts the AD curve by $\bar{a}$ at every inflation rate, and the economy moves along the AS curve, which has not moved yet. A positive shock raises both output and inflation, each by half the shock with the baseline numbers; a negative one lowers both, and at zero nothing changes.

**Source.** Adapted from Fig. 13.12 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 13: Dynamics as the AS curve shifts

*Move the slider to step through the periods of the boom; watch the AS curve shift up as the economy climbs AD₁ from B toward C.*

$$
\pi_t = \pi_{t-1} + \bar{\nu}\tilde{Y}_t
$$

$$
\tilde{Y}_t = \bar{a} - \bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

$$
\pi^c = \bar{\pi} + \frac{\bar{a}}{\bar{b}\bar{m}}
$$

where the demand shock $\bar{a} = 2$ lasts from period 1 to 9, and $\pi^c$ is inflation at point C, where output is back at potential.

```python
#| echo: false
# Model app: Fig. 13: Dynamics as the AS curve shifts
```

**Takeaway.** Inflation at B is above what firms expected, so each period the AS curve shifts up and the economy climbs AD₁: inflation rises toward $\pi^c$ and output falls back toward potential, by smaller steps each period. Point C looks like a resting point, with output at potential and stable inflation, but it is not the steady state: inflation is above the target, and the boom will not last forever.

**Source.** Adapted from Fig. 13.13 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 14: The unraveling after the aggregate demand shock ends

*Move the slider to step through the periods after the boom ends; watch the jump from C to D, then the slide back to A.*

$$
\tilde{Y}_t = -\bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

$$
\pi_t - \bar{\pi} = \lambda\,(\pi_{t-1} - \bar{\pi})
$$

where the boom has ended after period 9, so $\bar{a} = 0$ again, and $\lambda = 1/(1 + \bar{b}\bar{m}\bar{\nu})$ = 1/2.

```python
#| echo: false
# Model app: Fig. 14: The unraveling after the aggregate demand shock ends
```

**Takeaway.** When the boom ends after period 9, the AD curve shifts back, and with output already at potential this works like a negative demand shock: the economy jumps from C to D, a recession with inflation still above the target. From D the recession pulls inflation down, the AS curve shifts down period by period, and both gaps halve each period on the way back to A.

**Source.** Adapted from Fig. 13.14 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 15: A positive aggregate demand shock: summary

*Move the slider to set the last period of the boom; panel (a) traces the loop, panel (b) the paths over time.*

$$
\pi^c = \bar{\pi} + \frac{\bar{a}}{\bar{b}\bar{m}}
$$

$$
\sum_{t=1}^{T}\tilde{Y}_t = \frac{\pi_T - \pi_0}{\bar{\nu}}
$$

where $\bar{a} = 2$ lasts from period 1 to the last period of the boom, $\pi^c$ is the inflation rate the boom drifts toward, and the sum runs over the whole episode.

```python
#| echo: false
# Model app: Fig. 15: A positive aggregate demand shock: summary
```

**Takeaway.** The economy runs a counterclockwise loop: a boom with rising inflation, a drift up toward $\pi^c$, a recession when the boom ends, and the slow way back to A. Inflation starts and ends at the target, so the boom and the later recession cancel and output sums to about zero whatever the boom's length; a shorter boom only makes the loop smaller and the inflation peak lower.

**Source.** Adapted from Fig. 13.15 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## 13.9 Modern monetary policy and inflation targeting

### Fig. 20: Costless disinflation by coordinating expectations

*Move the slider to change the credibility of the new target; watch the AS curve follow the AD curve down, and the recession at B shrink.*

$$
\pi_t = \pi^e_t + \bar{\nu}\tilde{Y}_t + \bar{o}
$$

$$
\pi^e = c\,\bar{\pi}' + (1 - c)\,\bar{\pi}
$$

$$
\tilde{Y}_1 = -\bar{b}\bar{m}\lambda\,(1 - c)(\bar{\pi} - \bar{\pi}')
$$

where $c$ is the credibility of the new target $\bar{\pi}' = 2$ percent, $\bar{\pi} = 4$ percent the old one, and $\lambda = 1/(1 + \bar{b}\bar{m}\bar{\nu})$.

```python
#| echo: false
# Model app: Fig. 20: Costless disinflation by coordinating expectations
```

**Takeaway.** The announcement moves the AD curve all the way down to the new target, and credibility decides how far expected inflation, and with it the AS curve, follows. The first-period recession shrinks in proportion to $1 - c$: with no credibility inflation falls only halfway, as under adaptive expectations, and with full credibility it drops straight to the new target with output at potential.

**Source.** Adapted from Fig. 13.20 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612) The credibility slider, which spans the cases between adaptive and fully credible expectations, is our addition.

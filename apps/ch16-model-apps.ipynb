## Utility and the desire to smooth consumption

### Fig. 1: Flow utility

*Move the slider to change consumption; watch the tangent, whose slope is the marginal utility, flatten as consumption rises.*

$$
u(c) = \log c
$$

$$
u'(c) = \frac{1}{c}
$$

where $c$ is consumption in one period, $u(c)$ the utility it gives, and $u'(c)$ the marginal utility, the slope of the curve.

**Takeaway.** Utility rises with every extra unit of consumption, but by less and less: the marginal utility $1/c$, the slope of the tangent, falls as consumption rises. One more unit therefore adds a lot when consumption is low and little when it is high, and this diminishing marginal utility is the root of the desire to smooth consumption.

### Fig. 2: The desire to smooth consumption

*Move the slider to change future consumption, with today's fixed at 1; watch the gap between the dot on the curve and the open circle.*

$$
\bar{c} = \frac{c_1 + c_2}{2}
$$

$$
u(\bar{c}) \ge \tfrac{1}{2}\left[u(c_1) + u(c_2)\right]
$$

where $c_1$ is consumption today, $c_2$ consumption in the future, $\bar{c}$ the smooth path with the same total, and $u(c) = \log c$.

**Takeaway.** Because the utility curve bends, the point on the curve above $\bar{c}$ lies above the midpoint of the dashed line whenever the two periods differ, so the smooth path gives more utility than the uneven one with the same total. The more uneven the path, the larger the gain from smoothing; when future consumption equals today's, the two paths coincide and the gain is zero.

**Source.** Adapted from Fig. 16.2 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

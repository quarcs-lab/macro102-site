## A.1 What is Google Colab?

A **notebook** is a document that mixes three things: text (like this paragraph), code, and the results
of that code (numbers, tables and charts). You read it from top to bottom. When you change a piece of code
and run it again, its result updates right below it. That makes notebooks a good way to learn, because the
explanation, the computation and the evidence sit side by side.

**Google Colab** (short for Colaboratory) is a free service from Google that runs notebooks inside your web
browser. The notebooks follow the open Jupyter format (`.ipynb` files), which is the standard in data science.
When you run a notebook in Colab, the code does not run on your laptop. It runs on a computer in one of
Google's data centres, called the **runtime**, and the results come back to your browser.

### Why this book uses Colab

- **Nothing to install.** Python and all the libraries this book needs (NumPy, pandas, Plotly and
  ipywidgets) are already installed in Colab. You only need a web browser and a Google account.
- **The same setup for everyone.** Every reader runs the same versions of the same tools. Code that works
  for the author works for you, on Windows, macOS, Linux or a Chromebook.
- **Live models.** In Colab, the model figures of the book get sliders and buttons. Move a slider and the
  model is solved again on the spot.
- **Easy to save and share.** You can save your own copy of any notebook in Google Drive, and share it with a
  classmate or an instructor with a link, much like a Google Doc.
- **Data straight from the source.** Each notebook reads its data files over the internet from the book's
  GitHub repository, so you never have to download or upload a data file.

### What to keep in mind

- **The runtime is temporary.** When you close the notebook, or leave it idle for a while, Colab disconnects
  the runtime and forgets every variable. Free sessions also have a maximum length of several hours (see
  the [Colab FAQ](https://research.google.com/colaboratory/faq.html)). This is not a problem for this book:
  click *Runtime → Run all* again and everything is rebuilt in a few seconds.
- **Your edits are not saved automatically** in a notebook you opened from GitHub. Save your own copy in
  Google Drive first (Section A.2, step 6).
- **Free resources are shared.** The free tier limits how many notebooks can run at the same time. Close a
  runtime properly when you are done (Section A.2, step 7).
- **You need to be online**, because the runtime and the data both live on the internet.

## A.4 Python fundamentals

Python is a programming language that reads almost like English. This section covers the handful of ideas
you need for the whole book. Every example comes from the economics of the book.

### A.4.1 Numbers and arithmetic

Python works like a calculator. `**` raises a number to a power, which is all you need for compound growth.

```python
print(7 + 3, 7 - 3, 7 * 3, 7 / 3)     # add, subtract, multiply, divide
print(7 // 3, 7 % 3)                 # whole-number division and its remainder
print(2 ** 10)                       # a power: 2 to the 10th
```

An economy that grows at 2 percent a year is multiplied by 1.02 every year. After 35 years it is
$1.02^{35}$ times as large:

```python
1.02 ** 35
```

It has almost exactly doubled. This is the *rule of 70* of Chapter 3 at work: at 2 percent a year, a
variable doubles in about 70/2 = 35 years.

### A.4.2 Variables

A **variable** is a name for a value. The `=` sign means "store the value on the right under the name on the
left". Names may contain letters, digits and underscores, but not spaces, and they are case-sensitive
(`GDP` and `gdp` are different names). In this book the names follow the notation of the textbook, such as
`s`, `d`, `A`, `k_star` and `y_star`.

```python
y_1990 = 47713      # U.S. GDP per person in 1990, 2022 dollars (Maddison data, Section A.5)
y_2022 = 75459      # the same in 2022
growth_factor = y_2022 / y_1990
growth_factor
```

From the growth factor we get the **average annual growth rate**: the rate $g$ that turns $y_{1990}$ into
$y_{2022}$ in 32 years, $y_{2022} = y_{1990}(1+g)^{32}$, so $g = (y_{2022}/y_{1990})^{1/32} - 1$.

```python
g = growth_factor ** (1 / 32) - 1
g
```

### A.4.3 Types of values and f-strings

Every value has a **type**. The four you will meet most are whole numbers (`int`), decimal numbers
(`float`), text (`str`, for *string*, written in quotes) and truth values (`bool`: `True` or `False`).

```python
country = "United States"           # a string
years = 2022 - 1990                 # an int
g = growth_factor ** (1 / years) - 1    # a float
is_rich = y_2022 > 50000            # a bool, the result of a comparison
print(type(country), type(years), type(g), type(is_rich))
```

> **Python note.** An **f-string** is a string with an `f` before the opening quote. Python replaces each
> `{...}` inside it with the value of the expression. After a colon you can set the format: `{g:.2%}` shows
> `g` as a percentage with 2 decimals, `{y:,.0f}` shows `y` with a thousands separator and no decimals.

```python
print(f"{country}: GDP per person grew from {y_1990:,.0f} to {y_2022:,.0f} dollars,")
print(f"or {g:.2%} a year on average over {years} years.")
```

### A.4.4 Lists and loops

A **list** holds several values in order, inside square brackets. Python counts positions from **0**, so
`countries[0]` is the first element and `countries[-1]` the last one.

```python
countries = ["United States", "Japan", "China"]
print(countries[0], "|", countries[-1], "|", len(countries), "countries")
countries.append("India")        # add an element at the end
countries
```

A **`for` loop** repeats the indented lines below it, once for each value. `range(1, 6)` gives the
whole numbers 1, 2, 3, 4 and 5. Note the colon at the end of the `for` line and the four spaces of
indentation: they tell Python which lines belong to the loop.

```python
# Compound growth, year by year: income grows by 2 percent each year
y = 100.0
for year in range(1, 6):
    y = y * 1.02
    print(f"year {year}: {y:.2f}")
```

Loops are how the book simulates models over time. Here is the capital accumulation equation of the Solow
model (Chapter 5), $k_{t+1} = k_t + s\,k_t^{1/3} - d\,k_t$, run for 50 years. Each year, capital per person
grows by investment and shrinks by depreciation.

```python
s, d = 0.2, 0.1        # investment rate and depreciation rate
k = 1.0                # capital per person in year 0
path = [k]             # a list that will hold the whole path
for t in range(50):
    k = k + s * k ** (1 / 3) - d * k
    path.append(k)
k_star = (s / d) ** 1.5                 # the steady state of this model, (s/d)^(3/2)
print(f"k after 10 years: {path[10]:.3f}, after 50 years: {path[50]:.3f}")
print(f"steady state k* = {k_star:.3f}")
```

Capital rises quickly at first and then slows down as it approaches the steady state: the *principle of
transition dynamics* of Chapter 5.

### A.4.5 Making decisions with `if`

`if`, `elif` (else if) and `else` run different lines depending on a condition. The comparisons are `>`,
`<`, `>=`, `<=`, `==` (equal; note the double sign) and `!=` (not equal).

```python
g_percent = 1.4                  # an average growth rate, in percent a year
if g_percent >= 3:
    label = "fast"
elif g_percent >= 1:
    label = "moderate"
else:
    label = "slow or negative"
print(f"Growth of {g_percent}% a year is {label}.")
```

### A.4.6 Dictionaries

A **dictionary** stores values under names (*keys*), inside curly braces. The book uses dictionaries for
sets of parameters and for colors.

```python
params = {"s": 0.2, "d": 0.1, "A": 1.0}
print(params["s"])               # look up a value by its key
params["s"] = 0.3                # change it
for key in params:
    print(key, "=", params[key])
```

### A.4.7 Functions

A **function** packs a few lines of code under a name, so you can reuse them. `def` starts the definition,
the names in parentheses are its *arguments*, and `return` sends back the result. An argument with a
default value (`A=1.0`) can be left out when you call the function. The text in triple quotes, the
*docstring*, says what the function does. In the chapters each model equation becomes one short function
like this one, the production function of Chapter 4.

```python
def output(k, A=1.0):
    """Output per person in the production model of Chapter 4: y = A k^(1/3)."""
    return A * k ** (1 / 3)

print(output(8))           # k = 8 and the default A = 1
print(output(8, A=2.0))    # the same capital with twice the productivity
```

A second function turns the rule of 70 into code and uses a loop to try several growth rates.

```python
def years_to_double(g_percent):
    """Rule of 70 (Chapter 3): something growing at g percent a year doubles in about 70/g years."""
    return 70 / g_percent

for g_percent in [1, 2, 3.5, 7]:
    print(f"growth of {g_percent}% a year: doubles in about {years_to_double(g_percent):.0f} years")
```

### A.4.8 Libraries and errors

A **library** (or *package*) is a collection of functions that someone else has written. `import` loads
one. The built-in `math` library, for example, has the natural logarithm, which gives the exact doubling
time $\ln 2 / \ln(1+g)$ to compare with the rule of 70. The rule is very close at low growth rates and a
little off at high ones.

```python
import math
for g_percent in [2, 7]:
    exact = math.log(2) / math.log(1 + g_percent / 100)
    rule = years_to_double(g_percent)
    print(f"At {g_percent}% a year: exactly {exact:.1f} years; rule of 70: {rule:.1f} years")
```

Everyone writes code with mistakes. When Python cannot run a line, it stops and prints an **error message**
(a *traceback*). Read it from the **bottom**: the last line names the kind of error and what went wrong, and
an arrow `---->` points to the line that failed. Screenshot A.11 shows a typo in a variable name, which gives a
`NameError`. Colab may also offer an *Explain error* button that asks Google's AI assistant for an
explanation. It can help, but read its answer critically.

You can also catch an error yourself with `try ... except`. Python tries the indented lines under `try`; if
they fail with the named error, it runs the lines under `except` instead of stopping. The setup cell of every
chapter uses this to run one line that only works in Colab.

```python
try:
    "GDP in " + 2025          # text plus a number: Python refuses to guess what you mean
except TypeError as error:
    print("TypeError:", error)
print("GDP in " + str(2025))    # the fix: turn the number into text first
```

```python
# @title Helper for the website versions (run this cell; you do not need to read it)
def slider_figure(make_figure, name, values, start, label=None):
    """Precompute make_figure(name=value) for every value and join the results with one slider.

    The website cannot run Python, so each interactive figure also gets this precomputed version.
    """
    figures = [make_figure(**{name: v}) for v in values]
    first = values.index(start)
    fig = go.Figure(layout=figures[first].layout)
    owner = []                                  # which precomputed figure each trace comes from
    for i, f in enumerate(figures):
        for trace in f.data:
            trace.visible = (i == first)
            fig.add_trace(trace)
            owner.append(i)
    steps = []
    for i, f in enumerate(figures):
        layout = {"title": f.layout.title.to_plotly_json(),
                  "annotations": [a.to_plotly_json() for a in f.layout.annotations],
                  "shapes": [s.to_plotly_json() for s in f.layout.shapes]}
        visible = [o == i for o in owner]
        steps.append(dict(method="update", label=str(values[i]), args=[{"visible": visible}, layout]))
    fig.update_layout(sliders=[dict(active=first, steps=steps, pad=dict(t=60),
                                    currentvalue=dict(prefix=(label or name) + " = "))])
    return fig
```

### A.5.2 NumPy: math on whole arrays

A NumPy **array** is like a list of numbers, but you can do math on all its elements at once, without a

### A.5.6 ipywidgets: sliders and menus

**ipywidgets** turns the arguments of a function into controls. `interact(f, ...)` draws one control for
each argument and calls `f` again every time you move one. This is how every model figure in the book becomes
interactive: we write a function that returns a figure, and hand it to `interact`.

```python
def production_figure(A=1.0, alpha=1 / 3):
    """Production function y = A k^alpha (Chapter 4), with fixed axes so changes are visible."""
    k = np.linspace(0, 10, 200)
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=k, y=k ** (1 / 3), mode="lines", name="Baseline: A = 1, α = 1/3",
                             line=dict(color=GRAY, dash="dash")))
    fig.add_trace(go.Scatter(x=k, y=A * k ** alpha, mode="lines", name="Your production function",
                             line=dict(color=TEAL, width=3)))
    fig.update_layout(title=f"Output per person at k = 8: y = {A * 8 ** alpha:.2f} (baseline 2.00)",
                      xaxis=dict(title="Capital per person, k", range=[0, 10],
                                 tickvals=[2, 4, 6, 8, 10]),
                      yaxis=dict(title="Output per person, y", range=[0, 7]))
    return fig
```

> **Python note.** `FloatSlider` makes a slider for decimal numbers, from `min` to `max` in steps of
> `step`, starting at `value`. `Dropdown` makes a menu; each option is a pair `("label shown", value)`.
> `style=WIDE` shows the whole description. The `;` at the end hides an extra line of text.

```python
A_slider = FloatSlider(value=1.0, min=0.5, max=2.0, step=0.1,
                       description="productivity A", style=WIDE)
alpha_menu = Dropdown(options=[("1/4", 0.25), ("1/3", 1 / 3), ("1/2", 0.5)], value=1 / 3,
                      description="capital exponent α", style=WIDE)
interact(production_figure, A=A_slider, alpha=alpha_menu);
```

```python
# @title Website version: the production function with a precomputed slider for A
slider_figure(production_figure, "A", [0.5, 1.0, 1.5, 2.0], start=1.0, label="productivity A")
```

**Source.** Adapted from the production model of Chapter 4 in Jones (2024),
[*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

**What to notice.** Doubling $\bar{A}$ doubles output at every level of capital. A larger exponent $\alpha$
makes the curve bend less, so diminishing returns to capital set in more slowly.

## A.6 Putting it together: which economies grew fastest after 1990?

Our mini project uses every tool of this appendix. The question: which economies raised their GDP per
person fastest between 1990 and 2023, and which fell behind? The data come from the Penn World Table, with
one row per economy and year from 1950 to 2023.

**Step 1: load and look.**

```python
pwt = pd.read_csv(DATA_URL + "pwt_income.csv")
print(pwt.shape, "|", pwt["country"].nunique(), "economies")
print("years:", pwt["year"].min(), "to", pwt["year"].max())
pwt.head()
```

**Step 2: one row per economy with its 1990 and 2023 values.** We filter each year, keep the columns we need
and `merge` the two tables on the country code. `suffixes` tells pandas how to rename the two `gdp_pc`
columns. Then we apply the growth formula to whole columns at once.

```python
start = pwt[pwt["year"] == 1990][["iso3", "country", "gdp_pc"]]
end = pwt[pwt["year"] == 2023][["iso3", "gdp_pc"]]
both = start.merge(end, on="iso3", suffixes=("_1990", "_2023"))
both["growth"] = ((both["gdp_pc_2023"] / both["gdp_pc_1990"]) ** (1 / 33) - 1) * 100
print(len(both), "economies have data in both years")
both.sort_values("growth").head(3)
```

**Step 3: look at your data.** The slowest "grower" is Venezuela, at about −13 percent a year. As in the
chapters, that number is not plausible: hyperinflation broke the PWT price data after 2017. We drop it and
then list the five fastest and five slowest economies, adding the doubling time from our rule-of-70 function.

```python
both = both[both["iso3"] != "VEN"]
fastest = both.nlargest(5, "growth")
slowest = both.nsmallest(5, "growth")
fastest["doubles_in_years"] = years_to_double(fastest["growth"]).round(0)
columns = ["country", "gdp_pc_1990", "gdp_pc_2023", "growth"]
print("Fastest growth, 1990-2023 (percent a year):")
print(fastest[columns + ["doubles_in_years"]].round(1).to_string(index=False))
print("\nSlowest growth:")
print(slowest[columns].round(1).to_string(index=False))
print(f"\n{(both['growth'] < 0).sum()} of {len(both)} economies were poorer in 2023 than in 1990.")
```

**What to notice.**

- China grew at 7.6 percent a year: at that pace income doubles about every nine years, and it rose more
  than tenfold in 33 years.
- Guyana tops the list because of oil: production from its offshore oil fields began in 2019. Fast growth
  of GDP per person can come from a resource boom, not only from capital and ideas.
- Twelve economies were poorer in 2023 than in 1990, among them Zimbabwe, Tajikistan and the United Arab
  Emirates. In the Emirates, total GDP grew about 3.7 times, but the population grew 5.6 times, from 1.9 to
  10.6 million, so GDP *per person* fell.

**Step 4: an interactive chart.** Our last function draws GDP per person over time for any economies you
pick, on a ratio scale. In Colab, hold Ctrl (⌘ on a Mac) and click names in the list to choose several.

```python
def growth_figure(names=("China", "United States", "Zimbabwe")):
    """GDP per person since 1950 for the chosen economies, on a ratio scale."""
    chosen = pwt[pwt["country"].isin(names)]
    fig = px.line(chosen, x="year", y="gdp_pc", color="country", log_y=True,
                  range_x=[1950, 2023], range_y=[80, 400000],     # every value fits
                  labels={"year": "Year", "gdp_pc": "GDP per person (2021 dollars, ratio scale)",
                          "country": ""},
                  title=f"GDP per person since 1950: {len(names)} economies")
    fig.update_yaxes(tickvals=[100, 300, 1000, 3000, 10000, 30000, 100000, 300000])
    return fig
```

## The short-run model at a glance

### Fig. 1: The structure of the short-run model

*Move the slider to set the nominal interest rate, with inflation stuck at 2 percent; watch each circle of the chain respond.*

$$
R_t = i_t - \pi_t
$$

$$
\tilde{Y}_t = \bar{a} - \bar{b}\,(R_t - \bar{r})
$$

$$
\Delta\pi_t = \bar{\nu}\tilde{Y}_t + \bar{o}
$$

where $i_t$ is the nominal rate, $\pi_t$ sticky inflation (2 percent here), $\bar{r}$ the marginal product of capital and $\bar{\nu}$ the slope of the Phillips curve.

```python
#| echo: false
# Model app: Fig. 1: The structure of the short-run model
```

**Takeaway.** Because inflation is sticky, the nominal rate moves the real rate one for one, the real rate sets short-run output through the IS curve, and output sets the change in inflation. A rate above 7 percent puts the real rate above $\bar{r}$, so output falls below potential and inflation starts to fall; a rate below 7 percent does the opposite, and at 7 percent output is at potential and inflation stays put.

**Source.** Adapted from Fig. 12.1 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## The MP curve and the IS/MP diagram

### Fig. 3: The MP curve in the IS/MP diagram

*Move the slider to change the demand shock; the MP line stays at 5 percent, so watch where it crosses the IS curve.*

$$
R_t = i_t - \pi_t
$$

$$
\tilde{Y}_t = \bar{a} - \bar{b}\,(R_t - \bar{r})
$$

where the MP line is the real rate $R_t$ the central bank sets, here $\bar{r} = 5$ percent, $\bar{a}$ is the demand shock and $\bar{b}$ the sensitivity of investment to the rate.

```python
#| echo: false
# Model app: Fig. 3: The MP curve in the IS/MP diagram
```

**Takeaway.** The central bank chooses the height of the MP line; the IS curve then decides how much output that rate brings. With the rate held at $\bar{r}$, a positive demand shock shifts the IS curve right and raises output one for one with the shock, a negative one shifts it left into a recession, and with no shock the economy sits at potential, point A.

**Source.** Adapted from Fig. 12.3 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 4: Raising the interest rate in the IS/MP diagram

*Move the slider to set the new real interest rate; the dashed line is the old MP line, and the arrow shows the fall in output.*

$$
\tilde{Y}_t = \bar{a} - \bar{b}\,(R_t - \bar{r})
$$

$$
\tilde{Y}' = -\bar{b}\,(R' - \bar{r})
$$

where $R'$ is the new real rate, $\bar{r} = 5$ percent the marginal product of capital and $\bar{b} = 2$ the sensitivity of investment to the rate.

```python
#| echo: false
# Model app: Fig. 4: Raising the interest rate in the IS/MP diagram
```

**Takeaway.** Only the MP line moves: a higher real rate is a movement along the IS curve, not a shift of it. Above $\bar{r}$, borrowing costs more than new capital earns, so investment falls and output drops by $\bar{b}$ points for every point on the rate; at 5 percent nothing changes.

**Source.** Adapted from Fig. 12.4 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 5: Stabilizing the economy after a housing bubble

*Move the slider to change the size of the demand shock; panel (a) shows the recession at B, panel (b) the rate cut that reaches C.*

$$
\tilde{Y}_t = \bar{a}' - \bar{b}\,(R_t - \bar{r})
$$

$$
R' = \bar{r} + \frac{\bar{a}'}{\bar{b}}
$$

where $\bar{a}' < 0$ is the demand shock after the bust and $R'$ the real rate that brings output back to potential.

```python
#| echo: false
# Model app: Fig. 5: Stabilizing the economy after a housing bubble
```

**Takeaway.** The bust shifts the IS curve left, so at an unchanged rate output falls one for one with the shock. Cutting the real rate to $R'$ makes borrowing cheap enough to make up for the lost demand and takes the economy back to potential at C; the larger the shock, the larger the cut. In the model, monetary policy can fully offset any demand shock.

**Source.** Adapted from Fig. 12.5 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## The Phillips curve

### Fig. 7: The Phillips curve

*Move the slider to change short-run output and watch the economy slide along the Phillips curve; the title reads the change in inflation.*

$$
\pi_t = \pi_{t-1} + \bar{\nu}\tilde{Y}_t + \bar{o}
$$

$$
\Delta\pi_t = \bar{\nu}\tilde{Y}_t + \bar{o}
$$

where $\bar{\nu} = 0.5$ is the slope of the Phillips curve and $\bar{o}$ a price shock (zero here).

```python
#| echo: false
# Model app: Fig. 7: The Phillips curve
```

**Takeaway.** Short-run output sets the change in inflation, not its level. Output above potential raises inflation by $\bar{\nu}$ points for each point of boom, output below potential lowers it, and at potential (point A) inflation stays where it was. Every move of the slider is a movement along the curve, which itself never shifts.

**Source.** Adapted from Fig. 12.7 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 8: An oil price increase

*Move the slider to change the price shock; the dashed line is the old Phillips curve, and point A shows inflation at potential.*

$$
\Delta\pi_t = \bar{\nu}\tilde{Y}_t + \bar{o}
$$

$$
\Delta\pi_t = 0 \quad\Longrightarrow\quad \tilde{Y}_t = -\frac{\bar{o}}{\bar{\nu}}
$$

where $\bar{o}$ is the price shock and $\bar{\nu} = 0.5$ the slope of the Phillips curve; output stays at potential.

```python
#| echo: false
# Model app: Fig. 8: An oil price increase
```

**Takeaway.** A price shock shifts the whole Phillips curve by $\bar{o}$, so inflation changes even with output at potential: cost-push inflation. An oil price rise shifts the curve up, and holding inflation steady would then take a recession; a negative shock shifts it down, and at zero the curve does not move.

**Source.** Adapted from Fig. 12.8 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## Using the short-run model: disinflation and the Great Inflation

### Fig. 10: Tightening monetary policy

*Move the slider to set the new nominal rate, with inflation stuck at 10 percent; watch the MP line rise and output fall.*

$$
R_t = i_t - \pi_t
$$

$$
\tilde{Y}_t = \bar{a} - \bar{b}\,(R_t - \bar{r})
$$

where sticky inflation is $\pi_t = 10$ percent, as in 1979, so a nominal rate of 15 percent gives a real rate equal to $\bar{r} = 5$ percent.

```python
#| echo: false
# Model app: Fig. 10: Tightening monetary policy
```

**Takeaway.** Because inflation is sticky, every point the Fed adds to the nominal rate is a point on the real rate, so the MP line shifts up and the economy moves along the IS curve into a recession. At 15 percent the real rate equals $\bar{r}$ and nothing changes; the higher the rate above it, the deeper the recession.

**Source.** Adapted from Fig. 12.10 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 11: A recession and falling inflation

*Move the slider to set how deep the recession is, and watch point B slide down the Phillips curve; the title reads the fall in inflation.*

$$
\pi_t = \pi_{t-1} + \bar{\nu}\tilde{Y}_t + \bar{o}
$$

where $\pi_{t-1} = 10$ percent is inflation before the recession, $\bar{\nu} = 0.5$ the slope and $\bar{o} = 0$.

```python
#| echo: false
# Model app: Fig. 11: A recession and falling inflation
```

**Takeaway.** Firms facing weak demand raise their prices by less, so a recession lowers inflation by $\bar{\nu}$ points for each point of output lost: the deeper the recession, the faster inflation falls. With output at potential the economy stays at A and inflation does not move. Lower inflation is bought with lost output.

**Source.** Adapted from Fig. 12.11 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 12: The disinflation over time

*Move the slider to set how many years the Fed holds the rate high; watch inflation step down in panel (c) and the output lost in the title.*

$$
\pi_t = \pi_{t-1} + \bar{\nu}\tilde{Y}_t
$$

$$
\text{lost output} = \frac{\pi_{-1} - \pi^T}{\bar{\nu}}
$$

where $\pi_{-1} = 10$ percent is inflation before the disinflation, $\pi^T$ the inflation it ends at, and output is held 2 percent below potential while policy is tight.

```python
#| echo: false
# Model app: Fig. 12: The disinflation over time
```

**Takeaway.** With output 2 percent below potential, inflation falls one point a year for as long as the Fed holds the rate high; when the rate returns to $\bar{r}$, output goes back to potential and inflation stays at its new, lower level. Every point of inflation removed costs the same two percent-years of output, so a longer squeeze buys lower inflation at a proportionally higher cost.

**Source.** Adapted from Fig. 12.12 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

$$
\Delta\pi_t = \bar{\nu}\tilde{Y}_{\text{true}}
$$

where $g$ is potential growth: 3.5 percent on the believed path, the slider's rate on the true path after 1973; $\tilde{Y}_{\text{true}}$ is how far actual output is above true potential.

```python
#| echo: false
# Model app: Fig. 13: Mistaking a slowdown in potential for a recession
```

**Takeaway.** The slower true potential grows after 1973, the further actual output falls behind the believed path, and the bigger the recession policymakers think they see; at 3.5 percent there is no slowdown and the stimulus is plainly a boom. Actual output stays above true potential at every step, so the Phillips curve raises inflation every year after 1973, by the same amount whatever the slider says.

**Source.** Adapted from Fig. 12.13 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## How central banks control the nominal interest rate

### Fig. 18: How the central bank sets the nominal interest rate

*Move the slider to shift money demand; the money supply stays at 5, so watch the interest rate move from i* to i**.*

$$
M^d = m_0 - i
$$

$$
i^* = m_0 - M^s
$$

where $m_0$ is money demand at a zero interest rate and $M^s = 5$ the money supply.

```python
#| echo: false
# Model app: Fig. 18: How the central bank sets the nominal interest rate
```

**Takeaway.** The nominal interest rate is the price of holding money, and it settles where people want to hold exactly the money supplied. With the supply fixed, a rise in money demand raises the rate one for one and a fall lowers it; at $m_0 = 10$ the rate stays at 5 percent.

**Source.** Adapted from Fig. 12.18 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 19: Raising the nominal interest rate

*Move the slider to set the new money supply; the dashed line is the old one, and the arrow shows how the interest rate moves.*

$$
M^d = m_0 - i
$$

$$
i^{**} = m_0 - M^{s\prime}
$$

where $m_0 = 10$, the old money supply is 5 units and $M^{s\prime}$ is the new one.

```python
#| echo: false
# Model app: Fig. 19: Raising the nominal interest rate
```

**Takeaway.** Cutting the money supply leaves people wanting more money than there is, so the interest rate rises until demand falls back to the new supply: tight money means a higher rate. A supply above 5 units does the opposite and lowers the rate, and at 5 units nothing changes; each unit moves the rate by one point.

**Source.** Adapted from Fig. 12.19 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 20: Targeting the nominal interest rate

*Move the slider to shift money demand with the rate pegged at 5 percent; watch the quantity of money change while the rate stays put.*

$$
M^d = m_0 - i
$$

$$
M^s = m_0 - i^T
$$

where $i^T = 5$ percent is the target rate and $m_0$ money demand at a zero interest rate.

```python
#| echo: false
# Model app: Fig. 20: Targeting the nominal interest rate
```

**Takeaway.** To peg the rate, the central bank supplies whatever money is demanded at its target, so the money supply becomes a horizontal line. A rise in money demand is met with more money and a fall with less, while the interest rate, and with it the real rate and output, does not move at all; at $m_0 = 10$ nothing changes.

**Source.** Adapted from Fig. 12.20 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

$$

where $\bar{a}' < 0$ is the demand shock policymakers believe in; the true IS curve never moved.

**Takeaway.** Policymakers cut the real rate to offset a fall in demand that never happened, so on the true IS curve output overshoots potential by exactly the shock they believed in.

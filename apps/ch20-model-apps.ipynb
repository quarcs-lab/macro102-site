## The open economy in the short-run model

### Fig. 4: Increasing interest rates and the IS curve

*Move the slider to change how strongly net exports react to the interest rate; watch the IS curve turn flatter and point B move left.*

$$
\frac{NX_t}{\bar{Y}_t} = \bar{a}_{nx} - \bar{b}_{nx}\,(R_t - \bar{R}^w)
$$

$$
\tilde{Y}_t = \bar{a} - (\bar{b}_i + \bar{b}_{nx})\,(R_t - \bar{r})
$$

where $\tilde{Y}_t$ is short-run output, $R_t$ the real interest rate, raised here from $\bar{r} = 5$ to 6 percent, and $\bar{b}_i = 1$ the sensitivity of investment.

**Takeaway.** A higher real interest rate lowers investment and, through a stronger dollar, net exports, so each point of interest lowers output by $\bar{b}_i + \bar{b}_{nx}$ percent. A larger net-export sensitivity turns the IS curve flatter around point A and moves B further left, so monetary policy gets more powerful. At zero only investment reacts, as in a closed economy.

**Source.** Adapted from Fig. 20.4 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 5: An increase in foreign interest rates

*Move the slider to step through the periods; watch the economy jump from A to B, climb toward C, drop to D and return.*

$$
\bar{a} = \bar{b}_{nx}\,(\bar{R}^w - \bar{r})
$$

$$
\tilde{Y}_t = \bar{a} - \bar{b}\bar{m}\,(\pi_t - \bar{\pi})
$$

$$
\pi_t = \pi_{t-1} + \bar{\nu}\tilde{Y}_t
$$

where the foreign real rate $\bar{R}^w$ is 2 points above $\bar{r}$ in periods 1 to 9, $\bar{m}$ is the policy response and $\bar{\nu}$ the slope of the Phillips curve.

**Takeaway.** A higher foreign rate weakens the dollar and raises net exports, a demand shock that shifts AD out in periods 1 to 9: output jumps above potential, then inflation climbs toward 4 percent as output returns to potential. When the foreign rate falls back in period 10, AD shifts back and the economy drops into a recession, then slides down AD to its starting point. The U.S. did nothing, yet its economy boomed and slumped.

**Source.** Adapted from Fig. 20.5 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## Exchange-rate regimes and the policy trilemma

### Fig. 7: The policy trilemma in open economies

*Move the slider to switch regimes after a 1-point rise in the foreign interest rate; watch which corner of the triangle is lost.*

$$
1 + i_t = (1+i^w)\,\frac{E_t}{E^e_{t+1}}
$$

$$
\%\Delta E = \eta\,(\Delta i - \Delta i^w)
$$

$$
i_t = i^w
$$

where $E$ is the exchange rate, $i$ and $i^w$ the interest rates at home and abroad, and $\eta = 5$ percent per point of interest gap; the last line holds under a credible peg.

**Takeaway.** Uncovered interest parity ties the exchange rate, the interest rate and free capital flows together, so a country can choose any two and the third follows. After the foreign rate rises by 1 point, a floating rate keeps the interest rate and lets the currency depreciate, a currency board holds the exchange rate by raising the interest rate although the economy did not need it, and capital controls keep both only by blocking capital flows.

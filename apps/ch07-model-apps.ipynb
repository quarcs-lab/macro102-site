## Supply and demand in the labor market

### Fig. 3: The labor market

*Move the slider to shift labor demand up or down; the dashed gray lines are the starting curves, and the title reads the new equilibrium.*

$$
w = \bar{a}_d - \bar{b}_d L, \qquad w = \bar{a}_s + \bar{b}_s L
$$

$$
L^* = \frac{\bar{a}_d - \bar{a}_s}{\bar{b}_d + \bar{b}_s}
$$

$$
w^* = \frac{\bar{b}_s\bar{a}_d + \bar{b}_d\bar{a}_s}{\bar{b}_d + \bar{b}_s}
$$

where $L$ is employment, $w$ the wage, $\bar{a}_d$ the demand intercept, $\bar{a}_s = 10$ the supply intercept and $\bar{b}_d = \bar{b}_s = 1$ the slopes.

**Takeaway.** A higher demand intercept means firms value every worker more, so the demand curve shifts up along the supply curve and the wage and employment both rise; a lower one lowers both, and at 100 nothing changes. A shift in demand always moves the wage and employment in the same direction, which is how the data reveal a demand shock.

**Source.** Adapted from Fig. 7.3 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 4: An income tax at rate τ

*Move the slider to change the tax rate; watch the economy move from A to B and the green bar, the tax wedge, grow.*

$$
(1-\tau)\,w = \bar{a}_s + \bar{b}_s L
$$

$$
L_\tau = \frac{(1-\tau)\bar{a}_d - \bar{a}_s}{(1-\tau)\bar{b}_d + \bar{b}_s}
$$

$$
w_\tau = \bar{a}_d - \bar{b}_d L_\tau
$$

where $\tau$ is the tax rate, $w$ the wage firms pay and $(1-\tau)w$ the wage workers keep.

```python
#| echo: false
# Model app: Fig. 4: An income tax at rate τ
```

**Takeaway.** Workers decide how much to work from the wage they keep, so the tax shifts labor supply up and to the left, and the economy moves from A to B: employment falls, the wage firms pay rises and the wage workers keep falls. The gap between the two wages, the tax wedge, widens as the rate rises; at a zero rate there is no wedge and nothing changes.

**Source.** Adapted from Fig. 7.4 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

### Fig. 5: A reduction in labor demand

*Move the slider to change the fall in labor demand; the dashed line is the old demand curve, and the title reads the new equilibrium.*

$$
w = (\bar{a}_d - \Delta) - \bar{b}_d L
$$

$$
\Delta L^* = -\frac{\Delta}{\bar{b}_d + \bar{b}_s}
$$

$$
\Delta w^* = -\frac{\bar{b}_s\,\Delta}{\bar{b}_d + \bar{b}_s}
$$

where $\Delta$ is the fall in labor demand and $\bar{b}_d = \bar{b}_s = 1$ the slopes of the two curves.

```python
#| echo: false
# Model app: Fig. 5: A reduction in labor demand
```

**Takeaway.** When firms want fewer workers at every wage, the demand curve shifts down along the supply curve and the economy moves from A to B: the wage and employment both fall. With equal slopes the shock is split evenly between the two, a larger fall in demand lowers both by more, and with no fall nothing changes.

**Source.** Adapted from Fig. 7.5 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

$$

$$

where $\bar{w} = w^*$ is the old wage, which does not fall, and $\Delta$ the fall in labor demand.

### Fig. 6, the bathtub model: unemployment settles at its natural rate

*Move the slider to change the job finding rate; watch how fast unemployment falls and where it settles, at the dashed line.*

$$
\Delta U_{t+1} = \bar{s}E_t - \bar{f}U_t
$$

$$
u^* = \frac{\bar{s}}{\bar{f} + \bar{s}}
$$

$$
u_t = u^* + (1 - \bar{s} - \bar{f})^t\,(u_0 - u^*)
$$

where $U$ and $E$ are the unemployed and the employed, $\bar{s} = 0.01$ the monthly job separation rate, $\bar{f}$ the job finding rate and $u_0$ = 10 percent the starting rate.

```python
#| echo: false
# Model app: Fig. 6, the bathtub model: unemployment settles at its natural rate
```

**Takeaway.** Unemployment stops changing when the jobs lost each month equal the jobs found, at the natural rate. A higher job finding rate widens the drain: unemployment falls faster from its starting point and settles at a lower rate, while a lower one leaves it falling slowly toward a higher rate.

**Source.** Adapted from the bathtub model of unemployment (Section 7.4) in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)

## The rising return to education

### Fig. 8, present value: what 45 years of wages are worth today

*Hover over a bar to read what one year's 100,000-dollar wage, paid that many years from now, is worth today.*

$$
pdv = \frac{\text{future value}}{(1 + R)^T}
$$

$$
pdv = \text{payment}\times\frac{1 - \left(\frac{1}{1+R}\right)^n}{1 - \frac{1}{1+R}}
$$

where $R$ = 3 percent is the interest rate, $T$ the number of years until a payment arrives and $n$ = 45 the number of yearly payments.

```python
#| echo: false
# Model app: Fig. 8, present value: what 45 years of wages are worth today
```

**Takeaway.** A dollar paid in the future is worth less than a dollar today, because a smaller sum saved today would grow into it with interest, so each year's wage is divided by $(1+R)^T$ and the bars shrink with distance. Added up, 45 years of pay are worth only about 25 years' pay today: the value of a worker's human capital.

### Fig. 9: Understanding the rising return to education

*Move the slider to shift college demand out, with supply shifted out by 10; watch the college wage and the premium in the title.*

$$
\Delta L^* = \frac{x_d + x_s}{\bar{b}_d + \bar{b}_s}
$$

$$
\Delta w^* = \frac{\bar{b}_s\,x_d - \bar{b}_d\,x_s}{\bar{b}_d + \bar{b}_s}
$$

where $x_d$ and $x_s = 10$ are the outward shifts in college demand and supply, and $\bar{b}_d = 0.4$ and $\bar{b}_s = 0.6$ the slopes of the college curves.

```python
#| echo: false
# Model app: Fig. 9: Understanding the rising return to education
```

**Takeaway.** Both shifts raise college employment, but they pull the college wage in opposite directions: it rises only if demand shifts out by more than two thirds of the supply shift, the ratio of the slopes. With a small demand shift the extra supply wins and the premium falls; with a larger one demand wins, and more graduates go together with a higher premium, as in the data since 1980.

**Source.** Adapted from Fig. 7.9 in Jones (2024), [*Macroeconomics*, 6th ed.](https://wwnorton.com/books/9781324063612)
